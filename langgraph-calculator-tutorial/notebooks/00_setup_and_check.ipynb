# 00 · Setup and check

**Goal:** by the end you can open any lesson in this tutorial, know it is using the right Python and packages, and
(optionally) know your Gemini key works.

## The idea in plain English

Before we cook, we check the kitchen. This notebook doesn't teach LangGraph yet. It checks three things:

1. You are running the tutorial's own Python (the **virtual environment**, or *venv*), not some other one.
2. The packages the lessons need are installed, and you can see their versions.
3. If you have a Google Cloud API key, Gemini answers. If you don't, that's fine: lessons 01–18 never use it.

**Analogy:** a pilot's pre-flight checklist. Nothing exciting happens, but every later step depends on it.

## Step 1: Which Python is running?

A **kernel** is the program that runs your notebook's code. It should be the Python inside this project's `.venv`
folder. If the path printed below doesn't contain `.venv`, switch the kernel to **Python (langgraph-tutorial)**
(see the README).

````python
import sys

print("Python version:", sys.version.split()[0])
print("Python program:", sys.executable)
print("Using the project venv:", ".venv" in sys.executable)
````

## Step 2: Which package versions are installed?

`importlib.metadata.version` reads the version of an installed package without importing it. This tutorial targets
**LangGraph 1.x** and **LangChain 1.x**, so the first number of those two should be `1`.

````python
from importlib.metadata import version

packages = [
    "langgraph",
    "langchain",
    "langchain-core",
    "langchain-google-genai",
    "langgraph-checkpoint-sqlite",
    "python-dotenv",
]
for name in packages:
    print(f"{name:30} {version(name)}")
````

## Step 3: Do the imports the lessons use actually work?

Installing a package and importing from it are different things: names move between versions. This cell imports one
thing from each area the tutorial covers. No network is used. If it prints `All imports OK`, you are ready for
lessons 01–18.

````python
from langgraph.graph import StateGraph, START, END          # lessons 01-12
from langgraph.types import Command, Send, interrupt          # lessons 09, 10, 14
from langgraph.checkpoint.memory import InMemorySaver         # lesson 13
from langgraph.store.memory import InMemoryStore              # lesson 17
from langgraph.prebuilt import ToolNode                       # lesson 18
from langchain.tools import tool                              # lesson 18
from langchain.messages import AIMessage, HumanMessage        # lessons 18-19
from langchain.agents import create_agent                     # lesson 19

print("All imports OK")
````

## Step 4: Is there an API key?

Your key lives in a file called `.env` in the project folder (copy `.env.example` to make one). `load_dotenv` reads
that file into **environment variables**: settings that live outside your code, so the key is never written into a
notebook. We only print *whether* the key exists, never the key itself.

The notebook runs from the `notebooks/` folder, so the project folder is one level up.

````python
import os
from pathlib import Path
from dotenv import load_dotenv

PROJECT_ROOT = Path.cwd().parent
load_dotenv(PROJECT_ROOT / ".env")

HAS_KEY = bool(os.environ.get("GOOGLE_CLOUD_API_KEY"))
print("Project folder:", PROJECT_ROOT)
print("API key found:", HAS_KEY)
````

## Step 5 (optional): Say hello to Gemini

This is the only LLM call in this notebook. `llm_setup.py` in the project folder holds `get_llm()`, the one shared
helper in the whole tutorial. We add the project folder to `sys.path` (the list of places Python looks for
modules) so the notebook can import it.

**No key yet?** The cell skips the call and tells you so. You can carry on with lessons 01–18 and come back before
lesson 19.

````python
sys.path.insert(0, str(PROJECT_ROOT))

if HAS_KEY:
    from llm_setup import get_llm

    llm = get_llm()
    print(llm.invoke("Say hello in one short sentence.").text)  # quick connection test
else:
    print("No GOOGLE_CLOUD_API_KEY found: skipping the Gemini test.")
    print("That's fine. Lessons 01-18 run without it. See the README to add a key later.")
````

## What just happened

- **Step 1** printed the Python program running this notebook. It should be inside `.venv`.
- **Step 2** listed the installed versions. `langgraph` and `langchain` start with `1.`.
- **Step 3** imported everything later lessons need and printed `All imports OK`.
- **Step 4** looked for `GOOGLE_CLOUD_API_KEY` in `.env` and printed `True` or `False`.
- **Step 5** either got a one-sentence hello from Gemini or skipped politely. You may also see a warning about
  "automatic function calling (AFC)". It comes from Google's library, not your code, and is safe to ignore.

## Common mistakes

1. **Wrong kernel.** Step 2 or 3 fails with
   `ModuleNotFoundError: No module named 'langgraph'`.
   Fix: choose **Python (langgraph-tutorial)** in the kernel menu, then restart and run all.
2. **Key file in the wrong place or misspelled.** Step 4 prints `API key found: False`. The file must be named
   exactly `.env` (with the dot) and sit in the project folder, not in `notebooks/`. Restart the kernel after
   creating it.
3. **Key works but Vertex AI is not enabled.** Step 5 fails with an error containing `403` and `PERMISSION_DENIED`.
   Fix: enable the Vertex AI API for your Google Cloud project and check the key is allowed to call it.
